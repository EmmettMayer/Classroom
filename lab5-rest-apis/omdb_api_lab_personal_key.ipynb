### Exercise 1: Querying OMDb Search Endpoint
OMDb accepts query options via URL parameters. To perform a title search, pass parameter `s` (search keyword), `type` (e.g., `'movie'`), and your personal `apikey`.

**Task:**

````python
import requests
import pandas as pd

# Base OMDb API Endpoint
url = "http://www.omdbapi.com/"

# TODO 1: Enter your personal activated API key and set query parameters
params = {
    "s": "Inception",                     # Search title keyword
    "type": "movie",                     # Media type filter
    "apikey": "2c2ccd6" # <-- REPLACE WITH YOUR ACTIVATED KEY
}

# TODO 2: Make the GET request
response = requests.get(url, params=params)

# TODO 3: Check HTTP response status code with .status_code
if response.status_code == 200:
    print("Successfully connected to OMDb using your personal key!")
else:
    print(f"Connection failed: {response.status_code}")
````

### Exercise 3: Two-Stage Lookup (Detail Query by IMDb ID)
The search endpoint returns minimal summary fields (`Title`, `Year`, `imdbID`, `Type`, `Poster`). To fetch full details (e.g., plot, cast, director, ratings), perform a second request using parameter key `i` passed with the title's `imdbID` and your personal API key.

**Task:**
1. Extract the `imdbID` of the first item in `df`.
2. Define a new detail parameters dictionary containing `i` and `apikey` (pulling your personal key from `params["apikey"]`).
3. Send a second `GET` request to retrieve detail metadata.
4. Print key detail metrics (e.g., `Director`, `Actors`, `imdbRating`, `BoxOffice`).

````python
# TODO 1: Extract imdbID from the first row of df
target_imdb_id = df.loc[0, "imdbID"]

# Define detail query parameters using your personal key
detail_params = {
    "i": target_imdb_id,
    "apikey": params["apikey"]  # Reuses your personal key from Exercise 1
}

# TODO 2: Make the GET request for details
detail_response = requests.get(url, params=detail_params)

# TODO 3: Parse JSON detail payload and print specific fields
detail_data = detail_response.json()
print(f"Title: {detail_data.get('Title')}")
print(f"Director: {detail_data.get('Director')}")
print(f"IMDb Rating: {detail_data.get('imdbRating')}")
````

## Part 2: Critical Thinking & Short Answer Questions

### Question 1: API Parameter Mapping
In OMDb, what is the difference between passing the parameter key `s=` versus `t=` or `i=`? When would you use each? (OMDb has an API parameters section on its website)

*s= (Search): Performs a fuzzy keyword search and returns an array of matching title summaries. Use for search results/lists.

t= (Title): Returns full details for a single movie matched by exact title string. Use when you know the title but not the ID.

i= (IMDb ID): Returns full details for a single item by unique IMDb ID (e.g., tt1375666). Use for direct, unambiguous item lookups.*

---

### Question 2: Two-Stage API Pattern
Why do APIs like OMDb require a two-stage lookup pattern (`s` search $\rightarrow$ extract `imdbID` $\rightarrow$ `i` detail request) instead of returning full plot summaries and actor lists for every movie in the initial search results list?

*Bandwidth & Performance: Returning full details (plots, cast, ratings) for dozens of search results bloats payload size and degrades network/server performance.

Separation of Concerns: Stage 1 (s) allows quick filtering, while Stage 2 (i) retrieves heavy detail metadata only for specific selected items.*
